```python
from databricks_langchain import ChatDatabricks

chat_model = ChatDatabricks(
    endpoint="databricks-gpt-oss-120b",
    temperature=0.1,
    max_tokens=256
)
```

```python
response = chat_model.invoke("What is MLflow?")
```

```python
print("--- Content ---")
print(response.content)

print("\n" + "="*50 + "\n")

# 2. additional_kwargs
print("--- additional_kwargs ---")
print(response.additional_kwargs)

print("\n" + "="*50 + "\n")

# 3. response_metadata (Often contains detailed usage and model info)
print("--- response_metadata ---")
print(response.response_metadata)

# If you want to see specific keys from metadata like 'usage':
print("\n--- response_metadata['usage'] ---")
print(response.response_metadata.get('usage'))

print("\n" + "="*50 + "\n")

# 4. id (Unique run ID)
print("--- id ---")
print(response.id)
```
