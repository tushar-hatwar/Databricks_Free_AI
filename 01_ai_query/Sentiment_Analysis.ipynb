```python
%sql 
DROP TABLE customer_support_tickets;
DROP TABLE analyzed_tickets;
```

```python
%sql
CREATE TABLE IF NOT EXISTS customer_support_tickets (
  ticket_id INT,
  customer_message STRING
);
```

```python
%sql
-- Insert sample messy unstructured customer support messages
INSERT INTO customer_support_tickets VALUES
  (1, 'Hi, I ordered a laptop yesterday but it arrived broken. The screen is completely shattered. I am very upset and need a replacement ASAP!!!'),
  (2, 'Just wanted to say your product is amazing! Works perfectly. Very satisfied with my purchase. Great customer service too.'),
  (3, 'The app keeps crashing every time I try to upload photos... I have tried restarting my phone but it still doesnt work... pretty frustrated'),
  (4, 'Question about billing - why was I charged twice for my subscription? Please explain the charges on my account.'),
  (5, 'Delivery was late by 3 days but product quality is excellent. Overall okay but shipping needs improvement'),
  (6, 'URGENT: My account was hacked and someone made unauthorized purchases!!! Need immediate help to secure my account!!!'),
  (7, 'Can I get a refund for the defective headphones? Left ear stopped working after 2 weeks of normal use.'),
  (8, 'Love everything about your service! Been a customer for 5 years. Keep up the great work!'),
  (9, 'Installation instructions are confusing. The manual doesnt match the actual product. Need clear step-by-step guidance.'),
  (10, 'Just checking on my order status. When will item arrive? Order number xyz123.');
```

```python
%sql
-- View raw support tickets
SELECT * FROM customer_support_tickets;
```

```python
%sql
-- Analyze and extract insights from customer messages using ai_query
CREATE TABLE IF NOT EXISTS analyzed_tickets AS
SELECT 
  ticket_id,
  customer_message,
  -- Analyze sentiment
  ai_query(
    'databricks-llama-4-maverick',
    'Analyze the sentiment of this customer message: ' || customer_message || '. Return ONLY one word: Positive, Negative, or Neutral.'
  ) AS sentiment
FROM customer_support_tickets;
```

```python
%sql
-- View analyzed tickets with extracted insights
SELECT * FROM analyzed_tickets;
```
